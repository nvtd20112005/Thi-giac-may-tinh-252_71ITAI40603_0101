## Bài 5: Nhận xét về kích thước Filter

**Nhận xét:**

1. **Kích thước kernel càng lớn → ảnh càng mờ:** Kernel lớn hơn sẽ lấy trung bình từ nhiều pixel hơn, làm mất chi tiết.

2. **Box Filter:** Làm mờ đều, có thể tạo hiệu ứng "blocky" ở các cạnh.

3. **Gaussian Filter:** Làm mờ mượt hơn Box Filter vì các pixel gần tâm có trọng số cao hơn.

4. **Median Filter:** Hiệu quả nhất trong việc loại bỏ nhiễu salt-and-pepper vì nó chọn giá trị giữa, không bị ảnh hưởng bởi các giá trị cực đoan.

5. **Trade-off:** Kernel lớn loại bỏ nhiễu tốt hơn nhưng làm mất chi tiết ảnh. Cần chọn kích thước phù hợp.

## Bài 7: Filter bằng phép toán số học và logic
Xử lý hai ảnh để tách riêng quả bóng

```python
# a. Đọc và cắt hai ảnh về cùng kích thước 190x155
img1 = cv2.imread('hinh1.jpg')  # Đọc ảnh 1 (có quả bóng góc trái trên)
img2 = cv2.imread('hinh2.jpg')  # Đọc ảnh 2 (có quả bóng góc phải dưới)

# Cắt ảnh về kích thước 190x155
img1 = img1[:155, 1:191]  # Cắt ảnh 1: 155 hàng, 190 cột (từ cột 1 đến 191)
img2 = img2[:155, :190]  # Cắt ảnh 2: 155 hàng, 190 cột (từ cột 0 đến 190)

print('Kích thước ảnh 1:', img1.shape)  # In kích thước ảnh 1
print('Kích thước ảnh 2:', img2.shape)  # In kích thước ảnh 2
```

```python
# b. Chuyển ảnh sang ảnh âm bản (negative)
inv_img1 = cv2.bitwise_not(img1)  # Đảo ngược bit - tạo ảnh âm bản của ảnh 1
inv_img2 = cv2.bitwise_not(img2)  # Đảo ngược bit - tạo ảnh âm bản của ảnh 2

# Hiển thị ảnh gốc và ảnh âm bản
TungDuong.figure(figsize=(12, 8))  # Tạo figure

TungDuong.subplot(2, 2, 1)  # Vị trí 1
TungDuong.imshow(img1)  # Ảnh 1 gốc
TungDuong.title('Ảnh 1 gốc')  # Tiêu đề

TungDuong.subplot(2, 2, 2)  # Vị trí 2
TungDuong.imshow(inv_img1)  # Ảnh 1 âm bản
TungDuong.title('Ảnh 1 âm bản (Negative)')  # Tiêu đề

TungDuong.subplot(2, 2, 3)  # Vị trí 3
TungDuong.imshow(img2)  # Ảnh 2 gốc
TungDuong.title('Ảnh 2 gốc')  # Tiêu đề

TungDuong.subplot(2, 2, 4)  # Vị trí 4
TungDuong.imshow(inv_img2)  # Ảnh 2 âm bản
TungDuong.title('Ảnh 2 âm bản (Negative)')  # Tiêu đề

TungDuong.tight_layout()  # Căn chỉnh
TungDuong.show()  # Hiển thị
```

```python
# c. Xử lý để chỉ còn quả bóng
# Sử dụng phép XOR để tìm sự khác biệt giữa 2 ảnh
ball_xor = cv2.bitwise_xor(img1, img2)  # XOR hai ảnh để lấy phần khác nhau

# Sử dụng phép AND với ảnh âm bản
ball_and = cv2.bitwise_and(inv_img1, inv_img2)  # AND hai ảnh âm bản

# Hiển thị kết quả
TungDuong.figure(figsize=(12, 4))  # Tạo figure

TungDuong.subplot(1, 3, 1)  # Vị trí 1
TungDuong.imshow(ball_xor)  # Kết quả XOR
TungDuong.title('XOR hai ảnh')  # Tiêu đề

TungDuong.subplot(1, 3, 2)  # Vị trí 2
TungDuong.imshow(ball_and)  # Kết quả AND âm bản
TungDuong.title('AND hai ảnh âm bản')  # Tiêu đề

# Đảo ngược kết quả AND để có quả bóng đen trên nền trắng
ball_final = cv2.bitwise_not(ball_and)  # Đảo ngược kết quả
TungDuong.subplot(1, 3, 3)  # Vị trí 3
TungDuong.imshow(ball_final)  # Kết quả cuối cùng
TungDuong.title('Quả bóng (đảo ngược AND)')  # Tiêu đề

TungDuong.tight_layout()  # Căn chỉnh
TungDuong.show()  # Hiển thị
```

### Khắc phục vạch trắng còn xuất hiện:

Nếu vẫn còn xuất hiện các vạch trắng, có thể khắc phục bằng các cách sau:

1. **Morphological Operations:** Sử dụng mở (opening) hoặc đóng (closing) để loại bỏ nhiễu nhỏ.
2. **Threshold:** Áp dụng ngưỡng để làm rõ đen/trắng.
3. **Erosion/Dilation:** Xói mòn hoặc giãn nở để làm sạch cạnh.

```python
# Khắc phục vạch trắng bằng morphological operations
# Chuyển sang ảnh xám
gray_xor = cv2.cvtColor(ball_xor, cv2.COLOR_BGR2GRAY)  # Chuyển XOR sang xám

# Áp dụng threshold để phân ngưỡng
_, thresh = cv2.threshold(gray_xor, 127, 255, cv2.THRESH_BINARY)  # Ngưỡng 127

# Tạo kernel cho morphological operations
kernel = np.ones((3, 3), np.uint8)  # Kernel 3x3

# Áp dụng opening để loại bỏ nhiễu nhỏ
opened = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)  # Mở = xói mòn + giãn nở

# Áp dụng closing để lấp đầy lỗ nhỏ
closed = cv2.morphologyEx(opened, cv2.MORPH_CLOSE, kernel)  # Đóng = giãn nở + xói mòn

# Hiển thị kết quả
TungDuong.figure(figsize=(12, 4))  # Tạo figure

TungDuong.subplot(1, 3, 1)  # Vị trí 1
TungDuong.imshow(thresh, cmap='gray')  # Ảnh threshold
TungDuong.title('Threshold')  # Tiêu đề

TungDuong.subplot(1, 3, 2)  # Vị trí 2
TungDuong.imshow(opened, cmap='gray')  # Sau opening
TungDuong.title('Sau Opening')  # Tiêu đề

TungDuong.subplot(1, 3, 3)  # Vị trí 3
TungDuong.imshow(closed, cmap='gray')  # Sau closing
TungDuong.title('Sau Closing (Kết quả cuối)')  # Tiêu đề

TungDuong.tight_layout()  # Căn chỉnh
TungDuong.show()  # Hiển thị
```

### 4. Sobel Detector - Xác định biên ảnh

```python
# Đọc ảnh và chuyển sang grayscale
img_sobel = cv2.imread('hinh1.jpg')  # Đọc ảnh
gray_sobel = cv2.cvtColor(img_sobel, cv2.COLOR_BGR2GRAY)  # Chuyển sang ảnh xám

# Áp dụng Sobel detector
sobel_x = cv2.Sobel(gray_sobel, cv2.CV_64F, 1, 0, ksize=3)  # Sobel theo trục x (biên dọc)
sobel_y = cv2.Sobel(gray_sobel, cv2.CV_64F, 0, 1, ksize=3)  # Sobel theo trục y (biên ngang)
sobel_combined = cv2.magnitude(sobel_x, sobel_y)  # Kết hợp cả hai hướng

# Hiển thị kết quả
TungDuong.figure(figsize=(15, 5))  # Tạo figure

TungDuong.subplot(1, 4, 1)  # Vị trí 1
TungDuong.imshow(gray_sobel, cmap='gray')  # Ảnh gốc xám
TungDuong.title('Ảnh gốc (Gray)')  # Tiêu đề

TungDuong.subplot(1, 4, 2)  # Vị trí 2
TungDuong.imshow(np.abs(sobel_x), cmap='gray')  # Sobel X
TungDuong.title('Sobel X (Biên dọc)')  # Tiêu đề

TungDuong.subplot(1, 4, 3)  # Vị trí 3
TungDuong.imshow(np.abs(sobel_y), cmap='gray')  # Sobel Y
TungDuong.title('Sobel Y (Biên ngang)')  # Tiêu đề

TungDuong.subplot(1, 4, 4)  # Vị trí 4
TungDuong.imshow(sobel_combined, cmap='gray')  # Sobel kết hợp
TungDuong.title('Sobel Combined')  # Tiêu đề

TungDuong.tight_layout()  # Căn chỉnh
TungDuong.show()  # Hiển thị
```

### 6. Hough Transform - Xác định đường thẳng

```python
# Đọc ảnh và phát hiện cạnh
img_hough = cv2.imread('hinh1.jpg')  # Đọc ảnh
gray_hough = cv2.cvtColor(img_hough, cv2.COLOR_BGR2GRAY)  # Chuyển xám
edges = cv2.Canny(gray_hough, 50, 150)  # Phát hiện cạnh bằng Canny

# Áp dụng Hough Line Transform
lines = cv2.HoughLinesP(edges, 1, np.pi/180, threshold=100, minLineLength=50, maxLineGap=10)  # Phát hiện đường thẳng

# Vẽ các đường thẳng lên ảnh
img_lines = img_hough.copy()  # Copy ảnh gốc
if lines is not None:  # Nếu tìm thấy đường thẳng
    for line in lines:  # Duyệt qua từng đường
        x1, y1, x2, y2 = line[0]  # Lấy tọa độ 2 điểm đầu cuối
        cv2.line(img_lines, (x1, y1), (x2, y2), (0, 255, 0), 2)  # Vẽ đường màu xanh lá

# Hiển thị kết quả
TungDuong.figure(figsize=(12, 5))  # Tạo figure

TungDuong.subplot(1, 3, 1)  # Vị trí 1
TungDuong.imshow(cv2.cvtColor(img_hough, cv2.COLOR_BGR2RGB))  # Ảnh gốc
TungDuong.title('Ảnh gốc')  # Tiêu đề

TungDuong.subplot(1, 3, 2)  # Vị trí 2
TungDuong.imshow(edges, cmap='gray')  # Ảnh cạnh
TungDuong.title('Canny Edges')  # Tiêu đề

TungDuong.subplot(1, 3, 3)  # Vị trí 3
TungDuong.imshow(cv2.cvtColor(img_lines, cv2.COLOR_BGR2RGB))  # Ảnh với đường thẳng
TungDuong.title('Hough Lines')  # Tiêu đề

TungDuong.tight_layout()  # Căn chỉnh
TungDuong.show()  # Hiển thị
```
